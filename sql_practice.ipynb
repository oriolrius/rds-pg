# PostgreSQL SQL Practice

This notebook connects to a PostgreSQL RDS database for SQL practice.

```python
# Load environment variables and connect to database
import os
from dotenv import load_dotenv

load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT", "5432")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_NAME = os.getenv("DB_NAME")

print(f"Connecting to: {DB_HOST}:{DB_PORT}/{DB_NAME}")
```

```python
# Load SQL magic
%load_ext sql
%config SqlMagic.displaylimit = 50
```

```python
# Connect to PostgreSQL
%sql postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}
```

## Check Connection

Verify we're connected to PostgreSQL:

```python
%%sql
SELECT version();
```

## Create Sample Table

```python
%%sql
CREATE TABLE IF NOT EXISTS users (
    id SERIAL PRIMARY KEY,
    name VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    age INTEGER,
    city VARCHAR(50),
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

## Insert Sample Data

```python
%%sql
INSERT INTO users (name, email, age, city) VALUES
    ('Alice Smith', 'alice@example.com', 28, 'Barcelona'),
    ('Bob Johnson', 'bob@example.com', 35, 'Madrid'),
    ('Carol White', 'carol@example.com', 42, 'Barcelona'),
    ('David Brown', 'david@example.com', 31, 'Valencia'),
    ('Eva Garcia', 'eva@example.com', 25, 'Madrid')
ON CONFLICT (email) DO NOTHING;
```

# SQL Practice Exercises

## 1. SELECT Basics

```python
%%sql
-- Select all users
SELECT * FROM users;
```

```python
%%sql
-- Select specific columns
SELECT name, email FROM users;
```

## 2. WHERE Clause

```python
%%sql
-- Filter by city
SELECT * FROM users WHERE city = 'Barcelona';
```

```python
%%sql
-- Filter by age range
SELECT * FROM users WHERE age >= 30;
```

```python
%%sql
-- Multiple conditions
SELECT * FROM users WHERE city = 'Madrid' AND age < 40;
```

## 3. ORDER BY

```python
%%sql
-- Order by age descending
SELECT * FROM users ORDER BY age DESC;
```

## 4. Aggregate Functions

```python
%%sql
-- Count users
SELECT COUNT(*) as total_users FROM users;
```

```python
%%sql
-- Average age
SELECT AVG(age) as average_age FROM users;
```

```python
%%sql
-- Group by city
SELECT city, COUNT(*) as user_count, AVG(age) as avg_age
FROM users
GROUP BY city
ORDER BY user_count DESC;
```

## 5. UPDATE

```python
%%sql
-- Update a user's city
UPDATE users SET city = 'Sevilla' WHERE name = 'David Brown';
```

```python
%%sql
-- Verify the update
SELECT * FROM users WHERE name = 'David Brown';
```

## 6. LIKE Pattern Matching

```python
%%sql
-- Find users whose name contains 'a'
SELECT * FROM users WHERE name ILIKE '%a%';
```
